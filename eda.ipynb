Joining/merging data where applicable

```python
# Load and filter out cancellations/invalid entries for dimensional modeling
df = pd.read_csv('data.csv', encoding='latin1')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

valid_df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0) & (df['CustomerID'].notnull())].copy()

# Snapshot date for recency calculation (1 day past latest transaction)
snapshot_date = valid_df['InvoiceDate'].max() + pd.Timedelta(days=1)

# Build Customer Dimension
customer_dim = valid_df.groupby('CustomerID').agg(
    FirstPurchaseDate=('InvoiceDate', 'min'),
    LastPurchaseDate=('InvoiceDate', 'max'),
    TotalOrders=('InvoiceNo', 'nunique'),
    TotalSpend=('TotalAmount', 'sum'),
    PrimaryCountry=('Country', lambda x: x.mode().iloc[0] if not x.mode().empty else x.iloc[0])
).reset_index()

customer_dim['Recency_Days'] = (snapshot_date - customer_dim['LastPurchaseDate']).dt.days
customer_dim['CustomerTenure_Days'] = (snapshot_date - customer_dim['FirstPurchaseDate']).dt.days
customer_dim['CohortMonth'] = customer_dim['FirstPurchaseDate'].dt.to_period('M')

# Merge Customer Dimension back to Fact Table
fact_enriched = valid_df.merge(
    customer_dim[['CustomerID', 'CohortMonth', 'TotalOrders', 'Recency_Days']],
    on='CustomerID',
    how='left'
)
```

Categorizing data

```python
# 1. Define clean_df by filtering out returns, cancellations, and zero/negative prices
clean_df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)].copy()

# 2. Product Price Tiering (Line-Item Level)
price_bins = [0, 1.5, 5.0, 15.0, np.inf]
price_labels = ['Budget (<£1.5)', 'Mid-Range (£1.5-£5)', 'Premium (£5-£15)', 'Luxury (>£15)']
clean_df['Price_Tier'] = pd.cut(clean_df['UnitPrice'], bins=price_bins, labels=price_labels)

# 3. Geographic Regional Grouping
western_europe = ['Germany', 'France', 'EIRE', 'Spain', 'Netherlands', 'Belgium', 'Switzerland', 'Portugal']

def assign_region(country):
    if country == 'United Kingdom':
        return 'Domestic (UK)'
    elif country in western_europe:
        return 'Western Europe'
    return 'Rest of World'

clean_df['Region'] = clean_df['Country'].apply(assign_region)

# Verify the output
print(clean_df[['UnitPrice', 'Price_Tier', 'Country', 'Region']].head())
```

Aggregation

```python
invoice_agg = clean_df.groupby('InvoiceNo').agg(
    InvoiceDate=('InvoiceDate', 'first'),
    CustomerID=('CustomerID', 'first'),
    Country=('Country', 'first'),
    DistinctLines=('StockCode', 'nunique'),
    TotalQuantity=('Quantity', 'sum'),
    InvoiceTotal=('TotalAmount', 'sum')
).reset_index()

print("Overall AOV: £{:.2f}".format(invoice_agg['InvoiceTotal'].mean()))
```

```python
clean_df['YearMonth'] = clean_df['InvoiceDate'].dt.to_period('M')

monthly_summary = clean_df.groupby('YearMonth').agg(
    TotalRevenue=('TotalAmount', 'sum'),
    TotalOrders=('InvoiceNo', 'nunique'),
    ActiveCustomers=('CustomerID', 'nunique'),
    UnitsSold=('Quantity', 'sum')
).reset_index()

# Derived aggregated KPI
monthly_summary['AOV'] = (monthly_summary['TotalRevenue'] / monthly_summary['TotalOrders']).round(2)
```

```python
product_summary = clean_df.groupby(['StockCode', 'Description']).agg(
    TotalRevenue=('TotalAmount', 'sum'),
    TotalUnits=('Quantity', 'sum'),
    OrderFrequency=('InvoiceNo', 'nunique'),
    AvgPrice=('UnitPrice', 'mean')
).reset_index().sort_values(by='TotalRevenue', ascending=False)
```

Identifying unusual/outlier records

```python
# List of non-inventory administrative codes
admin_codes = ['AMAZONFEE', 'B', 'CRUK', 'D', 'DOT', 'M', 'POST', 'BANK CHARGES', 'PADS']

# 1. Flag administrative adjustments
df['IsAdminAdjustment'] = df['StockCode'].isin(admin_codes) | df['InvoiceNo'].str.startswith('A', na=False)

# 2. Flag damaged/discarded stock entries
df['IsInventoryLoss'] = (df['UnitPrice'] == 0) & (df['Quantity'] < 0)

# 3. Create clean Commercial Transactions view for Consumer Analytics (RFM / Retention)
clean_commercial = df[
    (~df['IsAdminAdjustment']) &
    (~df['IsInventoryLoss']) &
    (df['UnitPrice'] > 0) &
    (df['Quantity'] > 0) &
    (df['CustomerID'].notnull())
].copy()
```

Preparing an analysis-ready dataset

```python
def build_analysis_ready_dataset(filepath: str) -> pd.DataFrame:
    # 1. Load raw data
    raw_df = pd.read_csv(filepath, encoding='latin1')
    
    # 2. Define operational non-inventory codes
    admin_codes = ['AMAZONFEE', 'B', 'CRUK', 'D', 'DOT', 'M', 'POST', 'BANK CHARGES', 'PADS']
    
    # 3. Apply business filters
    clean_df = raw_df[
        raw_df['CustomerID'].notnull() &
        (raw_df['Quantity'] > 0) &
        (raw_df['UnitPrice'] > 0) &
        (~raw_df['StockCode'].isin(admin_codes)) &
        (~raw_df['InvoiceNo'].astype(str).str.startswith(('C', 'A')))
    ].copy()
    
    # 4. Type casting & string standardization
    clean_df['CustomerID'] = clean_df['CustomerID'].astype(int)
    clean_df['InvoiceDate'] = pd.to_datetime(clean_df['InvoiceDate'])
    clean_df['Description'] = clean_df['Description'].astype(str).str.strip().str.upper()
    
    # 5. Core calculated fields
    clean_df['LineTotal'] = (clean_df['Quantity'] * clean_df['UnitPrice']).round(2)
    clean_df['InvoiceYearMonth'] = clean_df['InvoiceDate'].dt.to_period('M')
    clean_df['DayOfWeek'] = clean_df['InvoiceDate'].dt.day_name()
    clean_df['Hour'] = clean_df['InvoiceDate'].dt.hour
    clean_df['IsUK'] = clean_df['Country'] == 'United Kingdom'
    
    # 6. Reorder columns for scannability
    ordered_cols = [
        'InvoiceNo', 'InvoiceDate', 'InvoiceYearMonth', 'DayOfWeek', 'Hour',
        'CustomerID', 'Country', 'IsUK',
        'StockCode', 'Description', 'Quantity', 'UnitPrice', 'LineTotal'
    ]
    
    return clean_df[ordered_cols].reset_index(drop=True)

# Run pipeline
df_clean = build_analysis_ready_dataset('data.csv')
```
